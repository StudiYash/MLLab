# **🩺 Diabetes Prediction**

This notebook performs a complete **data preprocessing + exploratory data analysis + model training** workflow on the **PIMA Diabetes Dataset**.  
The goal is to clean the dataset, treat medically-impossible zero-values, visualize feature relationships, and finally train multiple ML models to predict whether a patient is **diabetic (1)** or **non-diabetic (0)**.

---

## 🔎 **What This Code Does**

### **1️⃣ Loads and Inspects the Dataset**
- Reads the `diabetes.csv` file  
- Shows:
  - First 5 rows  
  - Dataset shape  
  - Data types and null values  
  - Summary statistics  
- Identifies incorrect zeros in medical features like:
  - Glucose  
  - Blood Pressure  
  - Skin Thickness  
  - Insulin  
  - BMI  

---

### **2️⃣ Data Cleaning & Imputation**
Many medical measurements cannot logically be zero.  
This block:

- Replaces all `0` values in key columns with `NaN`
- Displays missing value counts  
- Fills:
  - **Mean** for Glucose & BloodPressure  
  - **Median** for SkinThickness, Insulin, BMI  
- Produces a clean, medically valid dataset ready for training  

---

### **3️⃣ Exploratory Data Analysis (EDA)**
The notebook performs several visual analyses:

- Histograms of all features (before and after cleaning)
- Missing-value bar chart (`missingno`)
- Outcome distribution plot (number of diabetic vs non-diabetic cases)
- Insulin distribution + boxplot
- Correlation heatmap to understand how features interact  
  (e.g., Glucose is most strongly correlated with Outcome)

These visuals help understand the patterns and underlying structure of the dataset.

---

### **4️⃣ Feature Engineering & Train–Test Split**
- Separates dataset into:
  - **Features (X):** all medical measurements
  - **Target (y):** Outcome (0 or 1)
- Performs **stratified split** to preserve class ratios  
- Creates:
  - Training set (67%)  
  - Test set (33%)  

---

### **5️⃣ Machine Learning Models Trained**
This notebook trains **four different classifiers**, allowing a wide model comparison:

#### ✔ Random Forest Classifier  
- Measures training & test accuracy  
- Prints classification report  
- Shows feature importances (bar chart)

#### ✔ Decision Tree Classifier  
- Quick baseline model  
- Simple to interpret  
- Accuracy + confusion matrix included

#### ✔ XGBoost Classifier  
- Gradient-boosting model  
- Often performs best on structured medical data  
- Accuracy + detailed metrics printed

#### ✔ Support Vector Machine (SVM)  
- Uses RBF kernel  
- Effective for complex, non-linear decision boundaries  

Each model outputs:  
- Accuracy  
- Confusion matrix  
- Precision, Recall, F1-score

---

### **6️⃣ Model Serialization (Pickle)**
- Saves the best performing model (Random Forest) using `pickle`
- Reloads it
- Tests prediction accuracy after loading  
- Ensures the model can be deployed in production (API, app, etc.)

---

### **7️⃣ Final Predictions on New Patients**
The notebook predicts diabetes for two real patient test cases:

- Patient with metrics:  
  `[0, 137, 40, 35, 168, 43.1, 2.228, 33]`
- Patient with metrics:  
  `[10, 101, 76, 48, 180, 32.9, 0.171, 63]`

This demonstrates how to use the trained model for real-world inference.

---

## 🎯 **Final Outcome**

By the end of this notebook, the user will have:

- A fully cleaned and medically-validated dataset  
- Insightful visualizations about diabetes indicators  
- Four trained ML models  
- Performance comparison across algorithms  
- Feature importance insights  
- A pickled Random Forest model for deployment  
- Working prediction examples on new patient data  

```python
import warnings

import matplotlib.pyplot as plt
import missingno as msno
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# For Jupyter notebooks (safe to ignore in .py scripts)
%matplotlib inline

sns.set()

# ===================================================================
# Helper function to clearly separate sections in the notebook output
# ===================================================================

def section(title: str) -> None:
    """Print a clear, visually separated section header in the output."""
    print("\n" + "=" * 80)
    print(f"💠 {title}")
    print("=" * 80 + "\n")


# =========================
# 1. Load and Inspect Data
# =========================

section("1. Load Dataset")
diabetes_df = pd.read_csv("diabetes.csv")
print("Dataset loaded successfully with shape:", diabetes_df.shape)

section("2. Dataset Preview (First 5 Rows)")
print(diabetes_df.head())

section("3. Dataset Structure & Summary")
print("\n.info():")
print(diabetes_df.info())
print("\nDescriptive statistics (numerical features):")
print(diabetes_df.describe().T)

section("4. Missing Values (Raw Dataset)")
print(diabetes_df.isnull().sum())


# ======================================
# 2. Handle Invalid Zeros as Missing Data
# ======================================

section("5. Treat 0 as Missing in Selected Columns")

# Create a deep copy to preserve original

diabetes_df_copy = diabetes_df.copy(deep=True)

# Columns where 0 is considered invalid and treated as missing
cols_with_invalid_zero = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

# Replace 0 with NaN in selected columns
diabetes_df_copy[cols_with_invalid_zero] = diabetes_df_copy[cols_with_invalid_zero].replace(0, np.NaN)

print("Missing values per column AFTER replacing 0 with NaN:")
print(diabetes_df_copy.isnull().sum())


# =============================
# 3. Basic Distributions (Plots)
# =============================

section("6. Feature Distributions (Original Dataset with 0s)")
diabetes_df.hist(figsize=(20, 20))
plt.suptitle("Feature Distributions - Original Dataset", y=1.02)
plt.show()


section("7. Impute Missing Values in Cleaned Copy")
# Fill missing values with mean/median as appropriate

diabetes_df_copy["Glucose"].fillna(diabetes_df_copy["Glucose"].mean(), inplace=True)
diabetes_df_copy["BloodPressure"].fillna(diabetes_df_copy["BloodPressure"].mean(), inplace=True)
diabetes_df_copy["SkinThickness"].fillna(diabetes_df_copy["SkinThickness"].median(), inplace=True)
diabetes_df_copy["Insulin"].fillna(diabetes_df_copy["Insulin"].median(), inplace=True)
diabetes_df_copy["BMI"].fillna(diabetes_df_copy["BMI"].median(), inplace=True)

print("Missing values per column AFTER imputation:")
print(diabetes_df_copy.isnull().sum())

section("8. Feature Distributions (After Imputation)")
diabetes_df_copy.hist(figsize=(20, 20))
plt.suptitle("Feature Distributions - Cleaned Dataset", y=1.02)
plt.show()


section("9. Missing Value Visualization (Original Dataset)")
msno.bar(diabetes_df)
plt.title("Missing Values per Column (Original Dataset)")
plt.show()


# =============================
# 4. Outcome & Correlation Analysis
# =============================

section("10. Outcome Class Distribution")
print(diabetes_df["Outcome"].value_counts())

ax = diabetes_df["Outcome"].value_counts().plot(kind="bar")
ax.set_xlabel("Outcome")
ax.set_ylabel("Count")
ax.set_title("Outcome Distribution (0 = No Diabetes, 1 = Diabetes)")
plt.show()


section("11. Insulin Distribution & Boxplot (Original Dataset)")
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(diabetes_df["Insulin"], kde=True)
plt.title("Insulin Distribution")

plt.subplot(1, 2, 2)
plt.boxplot(diabetes_df["Insulin"].dropna())
plt.title("Insulin Boxplot")

plt.tight_layout()
plt.show()


section("12. Correlation Heatmap (Original Dataset)")
plt.figure(figsize=(12, 10))
sns.heatmap(diabetes_df.corr(), annot=True, cmap="RdYlGn")
plt.title("Correlation Heatmap")
plt.show()


# =============================
# 5. Feature Matrix and Label Vector
# =============================

section("13. Prepare Features (X) and Target (y) from Cleaned Dataset")

# Use the cleaned/imputed dataframe for modeling
X = diabetes_df_copy.drop("Outcome", axis=1)
y = diabetes_df_copy["Outcome"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)


# =============================
# 6. Train / Test Split
# =============================

section("14. Train / Test Split")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=7, stratify=y
)

print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)


# =============================
# 7. Random Forest Classifier
# =============================

section("15. Random Forest Classifier - Training & Evaluation")

rfc = RandomForestClassifier(n_estimators=200, random_state=7)
rfc.fit(X_train, y_train)

rfc_train_pred = rfc.predict(X_train)
train_acc = metrics.accuracy_score(y_train, rfc_train_pred)
print("Random Forest - Training Accuracy =", f"{train_acc:.4f}")

rfc_test_pred = rfc.predict(X_test)
test_acc = metrics.accuracy_score(y_test, rfc_test_pred)
print("Random Forest - Test Accuracy =", f"{test_acc:.4f}")

print("\nRandom Forest - Confusion Matrix:\n", confusion_matrix(y_test, rfc_test_pred))
print("\nRandom Forest - Classification Report:\n", classification_report(y_test, rfc_test_pred))

section("16. Random Forest Feature Importances")
plt.figure(figsize=(8, 6))
pd.Series(rfc.feature_importances_, index=X.columns).plot(kind="barh")
plt.title("Random Forest Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()


# =============================
# 8. Decision Tree Classifier
# =============================

section("17. Decision Tree Classifier - Training & Evaluation")

dtree = DecisionTreeClassifier(random_state=7)
dtree.fit(X_train, y_train)

dtree_pred = dtree.predict(X_test)
acc_dt = metrics.accuracy_score(y_test, dtree_pred)
print("Decision Tree - Test Accuracy =", f"{acc_dt:.4f}")
print("\nDecision Tree - Confusion Matrix:\n", confusion_matrix(y_test, dtree_pred))
print("\nDecision Tree - Classification Report:\n", classification_report(y_test, dtree_pred))


# =============================
# 9. XGBoost Classifier
# =============================

section("18. XGBoost Classifier - Training & Evaluation")

xgb_model = XGBClassifier(
    gamma=0,
    use_label_encoder=False,
    eval_metric="logloss",
    random_state=7
)
xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)
acc_xgb = metrics.accuracy_score(y_test, xgb_pred)
print("XGBoost - Test Accuracy =", f"{acc_xgb:.4f}")
print("\nXGBoost - Confusion Matrix:\n", confusion_matrix(y_test, xgb_pred))
print("\nXGBoost - Classification Report:\n", classification_report(y_test, xgb_pred))


# =============================
# 10. Support Vector Machine Classifier
# =============================

section("19. Support Vector Machine (SVM) - Training & Evaluation")

svc_model = SVC(random_state=7)
svc_model.fit(X_train, y_train)

svc_pred = svc_model.predict(X_test)
acc_svc = metrics.accuracy_score(y_test, svc_pred)
print("SVM - Test Accuracy =", f"{acc_svc:.4f}")
print("\nSVM - Confusion Matrix:\n", confusion_matrix(y_test, svc_pred))
print("\nSVM - Classification Report:\n", classification_report(y_test, svc_pred))


# =============================
# 11. Pickle the Best Model (Random Forest example)
# =============================

section("20. Serialize (Pickle) the Random Forest Model & Validate")

import pickle

# Serialize the RandomForest model
rfc_bytes = pickle.dumps(rfc)

# Load it back
rfc_from_pickle = pickle.loads(rfc_bytes)

# Verify loaded model works
pickle_pred = rfc_from_pickle.predict(X_test)
acc_pickle = metrics.accuracy_score(y_test, pickle_pred)
print("Pickled RF Model - Test Accuracy =", f"{acc_pickle:.4f}")


# =============================
# 12. Example Predictions on New Patients
# =============================

section("21. Example Predictions on New Patients")

example_1 = [[0, 137, 40, 35, 168, 43.1, 2.228, 33]]  # 4th patient
example_2 = [[10, 101, 76, 48, 180, 32.9, 0.171, 63]]  # 763rd patient

print("Example 1 (4th patient) - Predicted Outcome:", rfc.predict(example_1))
print("Example 2 (763rd patient) - Predicted Outcome:", rfc.predict(example_2))
```
